```python
import numpy as np




# Q1. Create a 1-D NumPy array with 5 elements


arr1 = np.array([10, 20, 30, 40, 50])

print("Q1: Original Array")
print(arr1)

print("\na) Add 2 to all elements:")
print(arr1 + 2)

print("\nb) Multiply all elements by 3:")
print(arr1 * 3)

print("\nc) Divide every element by 2:")
print(arr1 / 2)

```

```python


# Q2. Basic NumPy Array Operations



# Q2(a) Reverse the array
arr = np.array([1, 2, 3, 6, 4, 5])

print("\n\nQ2(a): Reversed Array")
print(arr[::-1])


# Q2(b)(i) Most frequent value and its indices
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])

unique_x, counts_x = np.unique(x, return_counts=True)
most_frequent_x = unique_x[np.argmax(counts_x)]
indices_x = np.where(x == most_frequent_x)[0]

print("\nQ2(b)(i): Array x")
print("Most frequent value:", most_frequent_x)
print("Indices:", indices_x)


# Q2(b)(ii) Most frequent value and its indices
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

unique_y, counts_y = np.unique(y, return_counts=True)
most_frequent_y = unique_y[np.argmax(counts_y)]
indices_y = np.where(y == most_frequent_y)[0]

print("\nQ2(b)(ii): Array y")
print("Most frequent value:", most_frequent_y)
print("Indices:", indices_y)

```

```python


# Q3. Access elements using row and column indices


arr_2d = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print("\n\nQ3: 2-D Array")
print(arr_2d)

# a) 1st row, 2nd column
print("\na) 1st row, 2nd column:", arr_2d[0, 1])
# b) 3rd row, 1st column
print("b) 3rd row, 1st column:", arr_2d[2, 0])



# Q4. Create an evenly spaced 1-D array using linspace()


# Replace 'your_name' with your actual name if required.
ucs420_your_name = np.linspace(10, 100, 25)

print("\n\nQ4: 1-D Array using linspace()")
print(ucs420_your_name)

print("\nDimensions:", ucs420_your_name.ndim)
print("Shape:", ucs420_your_name.shape)
print("Total elements:", ucs420_your_name.size)
print("Data type:", ucs420_your_name.dtype)
print("Total bytes:", ucs420_your_name.nbytes)

# Transpose using reshape()
transpose_reshape = ucs420_your_name.reshape(1, 25).T

print("\nTranspose using reshape():")
print(transpose_reshape)

# Can we do the same with T?
transpose_T = ucs420_your_name.reshape(1, 25).T

print("\nTranspose using T attribute:")
print(transpose_T)

print("\nNote: For a 1-D NumPy array, using .T directly does not change")
print("its shape. Reshaping it to 2-D first allows .T to produce a transpose.")
```

```python



# Q5. Create a 2-D array of 3 rows and 4 columns


ucs420_your_name_2d = np.array([
    [10, 20, 30, 40],
    [50, 60, 70, 80],
    [90, 15, 20, 35]
])

print("\n\nQ5: Original 2-D Array")
print(ucs420_your_name_2d)

print("\nMean:", np.mean(ucs420_your_name_2d))
print("Median:", np.median(ucs420_your_name_2d))
print("Maximum:", np.max(ucs420_your_name_2d))
print("Minimum:", np.min(ucs420_your_name_2d))
print("Unique elements:", np.unique(ucs420_your_name_2d))


# Reshape to 4 rows and 3 columns
reshaped_ucs420_your_name = ucs420_your_name_2d.reshape(4, 3)

print("\nReshaped array (4 x 3):")
print(reshaped_ucs420_your_name)


# Resize to 2 rows and 3 columns
resized_ucs420_your_name = np.resize(ucs420_your_name_2d, (2, 3))

print("\nResized array (2 x 3):")
print(resized_ucs420_your_name)
```
