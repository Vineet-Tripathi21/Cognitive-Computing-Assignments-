```python

ROLL_NUMBER = "1024170253"
NAME = "Vineet"
BRANCH = "COPC"
AGE = 20
CITY = "Lucknow"
CGPA = 8.91
```

```python
roll_digits = [int(d) for d in ROLL_NUMBER]


# Extract digits and multiply each by 10 to form list L
L = [d * 10 for d in roll_digits]
print("i.  L =", L)
```

```python
#Add two numbers - one using append(), one using insert()
L.append(999)                
print("ii. After append(999):", L)

L.insert(2, 555)              
print("ii. After insert(2, 555):", L)
```

```python
# iii. Remove two elements - one using remove(), one using pop()
L.remove(555)                
print("iii. After remove(555):", L)

popped = L.pop()             
print(f"iii. After pop() (removed {popped}):", L)
```

```python
# iv. Sort ascending, then descending
L.sort()
print("iv. Ascending:", L)
L.sort(reverse=True)
print("iv. Descending:", L)
```

```python
# v. Slicing - first three and last three elements
print("v. First three:", L[:3])
print("v. Last three:", L[-3:])

# vi. List comprehension - elements greater than the average
avg_L = sum(L) / len(L)
above_avg = [x for x in L if x > avg_L]
print(f"vi. Average = {avg_L:.2f}, elements above average:", above_avg)

```

```python
# Tuple made from the first 8 values of the ORIGINAL list L from Q1
original_L = [d * 10 for d in roll_digits]
scores = tuple(original_L[:8])
print("scores =", scores)

# i. Highest score + its index, lowest score + how many times it appears
highest = max(scores)
highest_index = scores.index(highest)
lowest = min(scores)
lowest_count = scores.count(lowest)
print(f"i. Highest = {highest} at index {highest_index}")
print(f"i. Lowest = {lowest}, appears {lowest_count} time(s)")
```

```python
# ii. Reverse the tuple, returned as a list
reversed_scores = list(reversed(scores))
print("ii. Reversed (as list):", reversed_scores)
# Tuples are immutable, so they have no in-place .reverse() method like lists do;
# reversed() must instead build a new sequence, which we convert to a list.

# iii. Ask user for a score, find its first index or report not present
user_score = int(input("iii. Enter a score to search for: "))
if user_score in scores:
    print(f"iii. {user_score} found at index {scores.index(user_score)}")
else:
    print(f"iii. {user_score} is not present in scores")
```

```python
# iv. Try to change an element directly - catch the error
try:
    scores[0] = 100
except TypeError as e:
    print("iv. Error raised:", e)
# Tuples are immutable, so item assignment is not allowed (TypeError);
# a list would allow this because lists are mutable.
```

```python
# v. Unpack into first, second, and remaining scores
first_score, second_score, *remaining_scores = scores
print("v. first:", first_score, "| second:", second_score, "| rest:", remaining_scores)

```

```python
import random
random.seed(int(ROLL_NUMBER))

# i. 100 random numbers between 100 and 900 inclusive
random_list = [random.randint(100, 900) for _ in range(100)]
print("i. Random list generated (first 10 shown):", random_list[:10], "...")

# ii. Count odd numbers
odd_count = len([n for n in random_list if n % 2 != 0])
print("ii. Odd count:", odd_count)
```

```python
# iii. Count even numbers
even_count = len([n for n in random_list if n % 2 == 0])
print("iii. Even count:", even_count)
```

```python
# iv. Count and list prime numbers
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True


primes = [n for n in random_list if is_prime(n)]
print("iv. Prime count:", len(primes))
print("iv. Primes found:", primes)
```

```python
# v. Most frequent number and its count
from collections import Counter
counts = Counter(random_list)
most_common_num, most_common_count = counts.most_common(1)[0]
print(f"v. Most frequent number: {most_common_num}, occurs {most_common_count} times")
```

```python
# Same 8 digits from roll number (BEFORE multiplying by 10)
digits_8 = roll_digits[:8]
A = {d * 7 for d in digits_8}
B = {d * 9 for d in digits_8}
print("A =", A)
print("B =", B)

# vi. Union
print("vi. Union (A | B):", A | B)

# vii. Intersection
print("vii. Intersection (A & B):", A & B)


```

```python
# viii. Difference each way
print("viii. A - B:", A.difference(B))
print("viii. B - A:", B.difference(A))

# ix. Symmetric difference
print("ix. Symmetric difference:", A.symmetric_difference(B))

# x. Subset / superset checks
print("x. Is A subset of B?", A.issubset(B))
print("x. Is B superset of A?", B.issuperset(A))

```

```python
# xi. Discard a user-input value from A
X = int(input("xi. Enter a value to discard from set A: "))
A.discard(X)
print("xi. A after discard:", A)
# discard() does nothing if the value isn't present, while remove() raises a KeyError;
# discard() is safer when you're not sure the value exists in the set.
```

```python
my_dict = {
    "name": NAME,
    "roll_no": ROLL_NUMBER,
    "branch": BRANCH,
    "age": AGE,
    "city": CITY,
}
print("Original dict:", my_dict)

# i. Rename "city" to "location" without hardcoding the value
my_dict["location"] = my_dict.pop("city")
print("i. After renaming city -> location:", my_dict)

# ii. Add "cgpa"
my_dict["cgpa"] = CGPA
print("ii. After adding cgpa:", my_dict)

# iii. Increase age by 1
my_dict["age"] += 1
print("iii. After incrementing age:", my_dict)
```

```python
# iv. Delete "branch" using pop() in one copy, del in another copy
dict_copy_1 = my_dict.copy()
popped_value = dict_copy_1.pop("branch")
print(f"iv. pop() removed branch and RETURNED its value: {popped_value}")
print("iv. dict_copy_1:", dict_copy_1)

dict_copy_2 = my_dict.copy()
del dict_copy_2["branch"]
print("iv. del removed branch but returns NOTHING (no value given back)")
print("iv. dict_copy_2:", dict_copy_2)

```

```python
# v. Iterate with .items()
print("v. Iterating with .items():")
for key, value in my_dict.items():
    print(f"   {key} -> {value}")

# vi. Safely check for a key that doesn't exist
if "email" in my_dict:
    print("vi. Email:", my_dict["email"])
else:
    print("vi. No 'email' key found - using fallback message instead of crashing")
```

```python
 #vii. Merge with a friend's dictionary
friend_dict = {
    "name": "Alex Friend",
    "roll_no": "9999999999",
    "branch": "ECE",
    "age": 21,
    "city": "Some Other City",
}
merged_dict = {**my_dict, **friend_dict}
print("vii. Merged dict:", merged_dict)


# viii. Dictionary comprehension - keep only string values
string_only_dict = {k: v for k, v in my_dict.items() if isinstance(v, str)}
print("viii. String-valued entries only:", string_only_dict)
```
