```python
# Assignment 5.1

# Question:
# WAP to find the maximum among three numbers.


# Solution:

a = int(input("Enter first number: "))
b = int(input("Enter second number: "))
c = int(input("Enter third number: "))

print("Largest =", max(a, b, c))

# Assignment 5.2

# Question:
# WAP to add all numbers divisible by both 7 and 9 from 1 to n.


# Solution:

n = int(input("Enter n: "))
total = 0

for i in range(1, n+1):
    if i % 7 == 0 and i % 9 == 0:
        total += i

print("Sum =", total)

# Assignment 5.3

# Question:
# WAP to add all prime numbers from 1 to n.


# Solution:

n = int(input("Enter n: "))
total = 0

for num in range(2, n+1):
    prime = True
    for i in range(2, int(num**0.5)+1):
        if num % i == 0:
            prime = False
            break
    if prime:
        total += num

print("Sum =", total)
```

```python
# Assignment 6.1

# Question:
# WAP using a function to add all odd numbers from 1 to n.


# Solution:

def add_odd(n):
    total = 0
    for i in range(1, n+1, 2):
        total += i
    return total

n = int(input("Enter n: "))
print("Sum =", add_odd(n))

# Assignment 6.2

# Question:
# WAP using a function to add all prime numbers from 1 to n.


# Solution:

def add_prime(n):
    total = 0
    for num in range(2, n+1):
        prime = True
        for i in range(2, int(num**0.5)+1):
            if num % i == 0:
                prime = False
                break
        if prime:
            total += num
    return total

n = int(input("Enter n: "))
print("Sum =", add_prime(n))
```
