## Q1. Sensor Readings

```python
import numpy as np

temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

# a. Add 2°C to every reading using vectorization
corrected_temperature = temperature + 2
print("Corrected temperatures:", corrected_temperature)

# b. Convert Celsius to Fahrenheit
fahrenheit = (9/5) * corrected_temperature + 32
print("Fahrenheit:", fahrenheit)

# c. Identify readings greater than 32°C
greater_than_32 = corrected_temperature[corrected_temperature > 32]
print("Readings greater than 32°C:", greater_than_32)

# d. Count readings greater than 32°C
count = np.sum(corrected_temperature > 32)
print("Count:", count)
```

## Q2. Daily Steps

```python
steps = np.array([
    [5000, 6200, 7100],
    [8000, 7500, 9000],
    [4500, 5100, 4800],
    [9000, 8500, 9500]
])

# a. Total steps
print("Total:", steps.sum())

# b. Mean number of steps
print("Mean:", steps.mean())

# c. Maximum and minimum
print("Maximum:", steps.max())
print("Minimum:", steps.min())

# d. Total steps for each day
print("Total for each day:", steps.sum(axis=0))

# e. Total steps for each user
print("Total for each user:", steps.sum(axis=1))

# f. Position of maximum value
position = np.unravel_index(np.argmax(steps), steps.shape)
print("Maximum position (0-based):", position)
```

## Q3. NumPy Slicing, Copying and Reshaping

```python
import numpy as np

# a. Create the original array
original = np.array([1, 2, 3, 4, 5, 6])

# b-c. Slice and modify it
subset = original[1:5]
subset[0] = 999

print("Original after modifying subset:", original)
print("Subset:", subset)

# d. Create a copy and modify it
copied = original[1:5].copy()
copied[0] = 500

print("Original after modifying copied array:", original)
print("Copied array:", copied)

# e. Create numbers from 1 to 12 and reshape into 3 x 4
matrix = np.arange(1, 13).reshape(3, 4)

print("3 x 4 matrix:")
print(matrix)

# f. Indexing and slicing
print("First row:", matrix[0])
print("Last row:", matrix[-1])
print("Second column:", matrix[:, 1])
print("Rows 1-2 and columns 2-3:")
print(matrix[0:2, 1:3])

# g. Flatten using flatten() and ravel()
flat1 = matrix.flatten()
flat2 = matrix.ravel()

print("flatten():", flat1)
print("ravel():", flat2)

# h. Modify ravel()
flat2[0] = 100
print("Matrix after changing ravel():")
print(matrix)

# i. Modify flatten()
flat1[1] = 200
print("Matrix after changing flatten():")
print(matrix)

# j. Array properties
print("Shape:", matrix.shape)
print("Dimensions:", matrix.ndim)
print("Size:", matrix.size)
print("Data type:", matrix.dtype)
```

## Q4. Cognitive-Assistive System and OLS

```python
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])

y = np.array([40, 65, 30, 85])

# a. Shape and dimensions
print("Shape:", X.shape)
print("Dimensions:", X.ndim)

# b. Transpose
print("\nX.T:")
print(X.T)

# c. Matrix product
XtX = X.T @ X
print("\nX.T @ X:")
print(XtX)

# d. Matrix inverse
XtX_inv = np.linalg.inv(XtX)
print("\nInverse of X.T @ X:")
print(XtX_inv)

# e. Ordinary Least Squares
beta = XtX_inv @ X.T @ y
print("\nOLS coefficients:")
print(beta)

# g. Prediction for a new user
new_user = np.array([5, 40, 7])
prediction = new_user @ beta

print("\nPredicted assistance score:", prediction)
```
